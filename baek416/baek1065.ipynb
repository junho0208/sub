```python
n = int(input())
def han_num(n):
    n_list = []
    for i in range(1,n+1):
        if i <= 99:
            n_list.append(i)
        else:
            k = str(i)
            dif = int(k[0]) - int(k[1])
            for j in range(1,len(k) - 1):
                if int(k[j]) - int(k[j+1]) == dif:
                    n_list.append(i)
    return len(n_list)
print(han_num(n))
```
