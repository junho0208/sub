```python
M = int(input())
N = int(input())

def prime_number(m,n):
    prime_list = []
    start = max(2,m)
    for i in range(start,n+1):
        p = True
        for j in range(2,i):
            if i % j == 0:
                p = False
                break
        if p:
            prime_list.append(i)
    return prime_list
if len(prime_number(M,N)) == 0:
    print(-1)
else:
    print(sum(prime_number(M,N)))
    print(min(prime_number(M,N)))
```
