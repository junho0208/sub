```python
n = int(input())
x = list(map(int, input().split()))
total = 0
for i in range(n):
    for j in range(n):
        total += abs(x[i] - x[j])
print(total)
#시간 초과
```

```python
n = int(input())
x = sorted(list(map(int, input().split())))
total = 0
for i in range(n):
    total += x[i] * (2 * i - n + 1)
print(total*2)
```
