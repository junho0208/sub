```python
T = int(input())
for i in range(T):
    s = input()
    VPS = []
    for j in s:
        if j == '(':
            VPS.append(j)
        elif j == ')':
            if not VPS:
                print('NO')
                break
            VPS.pop()
    else:
        if not VPS:
            print("YES")
        else:
            print("NO")
```
