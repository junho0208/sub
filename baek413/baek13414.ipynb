```python
import sys
input = sys.stdin.readline
K, L = map(int, input().split())

student_id = {}

for i in range(L):
    s = input().strip()
    student_id[s] = i

sorted_students = sorted(student_id.items(), key=lambda x: x[1])
for i in range(min(K, len(sorted_students))):
    print(sorted_students[i][0])
```
